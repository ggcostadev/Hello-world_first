```python
#Entrada de dados:
print('Aluno Edmundo - RA 764493')
print()
nota1 = float(input('Digite a nota da primeira prova (0 a 10): '))
nota2 = float(input('Digite a nota da segunda prova (0 a 10): '))
nota3 = float(input('Digite a nota da terceira prova (0 a 10): '))
nota4 = float(input('Digite a nota da quarta prova (0 a 10): '))
print()
#Processamento de dados:
media_das_notas = (nota1 + nota2 + nota3 + nota4)/4
print(f'Media final do aluno: {media_das_notas:.2f}')
print()
if media_das_notas < 5:
  print("Resultado: Reprovado")
else:
  print("Resultado: Aprovado")

```
